```python
import json
import os
```

```python
dir_path = "data"
for file_name in os.listdir(dir_path):
    file_path = os.path.join(dir_path, file_name)
    if os.path.isfile(file_path) and file_name.endswith('.json'):
        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f) 
            content = [msg.get('content', '').encode('latin1').decode('utf-8') for msg in data.get('messages', [])]
            with open("input.txt", "a", encoding="utf-8") as out_f:
                for s in content:
                    out_f.write(s + "\n")
```

```python
with open("input.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

lines.reverse()

with open("input.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)
```
